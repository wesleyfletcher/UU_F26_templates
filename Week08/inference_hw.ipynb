```python
import numpy as np
import pandas as pd
from scipy.stats import norm

rng = np.random.default_rng(12345)
boot = 2_000
```

# Problem 1

$\bar X = 52,       s = 10,        n = 100$.

$\hat{SE}(\bar X) = \frac{s}{\sqrt{n}} = \frac{10}{\sqrt{100}} = 1$

$z = \frac{\bar X - \mu}{\hat{SE}} = \frac{2}{1} = 2$

Two-tailed p-value for $z = 2$ is $p = 0.0455$.

Therefore, we can reject the null hypothesis for $\alpha = .10, .05$, but not for $\alpha=.01$.

95% CI = $\bar X \pm 1.96*\hat{SE} = (50.04, 53.96)$

# Problem 2

$\bar X = 11.5,       s = 3,        n = 36$.

$\hat{SE} = \frac{3}{\sqrt{36}} = 0.5$.

$12$ is only one standard error away from the sample mean, so it is a plausible population mean.

95% CI = $\bar X \pm 1.96*\hat{SE} = (10.52, 12.48)$

$z=\frac{11.5-12}{0.5} = -1$, so the p-value is $p=0.317$. Therefore we fail to reject the null hypothesis that $\mu = 12$ at $\alpha = .05$.

# Problem 3

$\bar X_A = 25,       s_A = 8,        n_A = 100$. \
$\bar X_B = 22,       s_B = 7,        n_B = 100$.

$\hat{SE}_A = \frac{8}{\sqrt{100}} = 0.8$. \
$\hat{SE}_B = \frac{7}{\sqrt{100}} = 0.7$.

$Z = \frac{\bar X_A - \bar X_B}{\sqrt{\hat{SE}_A^2 + \hat{SE}_B^2}} = \frac{3}{\sqrt{1.13}} = 2.82$

95% CI = $(\bar X_A - \bar X_B) \pm 1.96*\sqrt{\hat{SE}_A^2 + \hat{SE}_B^2} = (0.92, 5.08)$

# Problem 4

```python
ames_prices = pd.read_csv('./data/ames_prices.csv')
ames_prices['AirIndicator'] = (ames_prices['Central.Air'] == 'Y')
```

```python
n = len(ames_prices)
means = []

for i in range(boot):
    samp = ames_prices.sample(n, random_state=rng, replace=True)
    means.append(samp['AirIndicator'].mean())

means = np.array(means)
```

```python
MEAN_EST = means.mean()
SE = means.std(ddof=1)

print(f"MEAN ESTIMATE  = {MEAN_EST:.3f} \nSTANDARD ERROR = {SE:.3f}")
```

```python
CI_90 = (MEAN_EST - 1.645*SE, MEAN_EST + 1.645*SE)
CI_95 = (MEAN_EST - 1.96*SE,  MEAN_EST + 1.96*SE)
CI_99 = (MEAN_EST - 2.576*SE, MEAN_EST + 2.576*SE)

print(f"90% CI: ({CI_90[0]:.3f}, {CI_90[1]:.3f})")
print(f"95% CI: ({CI_95[0]:.3f}, {CI_95[1]:.3f})")
print(f"99% CI: ({CI_99[0]:.3f}, {CI_99[1]:.3f})")
```

```python
MEAN_NULL = .90
Z_SCORE = (MEAN_EST-MEAN_NULL)/(SE)
P_VALUE = 2*norm.sf(abs(Z_SCORE)) # 2-tailed

print(f"z = {Z_SCORE:.3f}")
print(f"p = {P_VALUE:.3f}")
```

For $\alpha = .10, .05, .01$, we reject the null hypothesis at all alphas. This matches with our confidence intervals, none of which contain 0.9. Based on our samples, we estimate that 93.3% of homes have central air. With 99% certainty, we can say that the true proportion is between 92.1% and 94.5%. There is less than a 1% chance that 90% is the true proportion.

# Problem 5

```python
ames_prices['arcsinh_area'] = np.arcsinh(ames_prices['area'])
ames_prices['arcsinh_price'] = np.arcsinh(ames_prices['price'])
```

```python
n = len(ames_prices)
rhos = []

for i in range(boot):
    samp = ames_prices.sample(n, random_state=rng, replace=True)
    rhos.append(samp[['arcsinh_area', 'arcsinh_price']].corr().iloc[0, 1])

rhos = np.array(rhos)
```

```python
RHO_EST = rhos.mean()
SE = rhos.std(ddof=1)

print(f"RHO ESTIMATE  = {RHO_EST:.3f} \nSTANDARD ERROR = {SE:.3f}")
```

```python
CI_90 = (RHO_EST - 1.645*SE, RHO_EST + 1.645*SE)
CI_95 = (RHO_EST - 1.96*SE,  RHO_EST + 1.96*SE)
CI_99 = (RHO_EST - 2.576*SE, RHO_EST + 2.576*SE)

print(f"90% CI: ({CI_90[0]:.3f}, {CI_90[1]:.3f})")
print(f"95% CI: ({CI_95[0]:.3f}, {CI_95[1]:.3f})")
print(f"99% CI: ({CI_99[0]:.3f}, {CI_99[1]:.3f})")
```

```python
RHO_NULL = 0
Z_SCORE = (RHO_EST-RHO_NULL)/(SE)
P_VALUE = 2*norm.sf(abs(Z_SCORE)) # 2-tailed

print(f"z = {Z_SCORE:.3f}")
print(f"p = {P_VALUE:.3f}")
```

For $\alpha = .10, .05, .01$, we reject the null hypothesis at all alphas. This matches with our confidence intervals, none of which contain 0. Based on our samples, we estimate that the pearson correlation between `arcsinh(area)` and `arcsinh(price)` is 0.723. With 99% certainty, we can say that the true correlation is between 0.697 and 0.749. There is a much less than 1% chance that 0 is the true correlation.

If we separated the instances of area and price, we would lose the relationship between area and price that is observed. This would create a much lower correlation than what the data suggests.

```python
CI_90 = (MEAN_EST - 1.645*SE, MEAN_EST + 1.645*SE)
CI_95 = (MEAN_EST - 1.96*SE,  MEAN_EST + 1.96*SE)
CI_99 = (MEAN_EST - 2.576*SE, MEAN_EST + 2.576*SE)

print(f"90% CI: ({CI_90[0]:.3f}, {CI_90[1]:.3f})")
print(f"95% CI: ({CI_95[0]:.3f}, {CI_95[1]:.3f})")
print(f"99% CI: ({CI_99[0]:.3f}, {CI_99[1]:.3f})")
```

```python
MEAN_NULL = 8
Z_SCORE = (MEAN_EST-MEAN_NULL)/(SE)
P_VALUE = 2*norm.sf(abs(Z_SCORE)) # 2-tailed

print(f"z = {Z_SCORE:.3f}")
print(f"p = {P_VALUE:.3f}")
```

For $\alpha = .10, .05, .01$, we reject the null hypothesis at all alphas. This matches with our confidence intervals, none of which contain 8. Based on our samples, we estimate that the average sleep time is 7.6 hours. With 99% certainty, we can say that the true time is between 7.566 and 7.678 hours. There is less than a 1% chance that 8 hours is the true duration mean.

While we can be almost certain the mean is not exactly 8 hours, it is still the closest whole number. If we were telling people how long they should sleep, telling someone 7.622 hours is not as simple to interpret as simply saying 8 hours. With statistical significance, we know that 8 is not the true mean, but for all practical purposes it might as well be 8 hours.
