```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
metabric = pd.read_csv('data/metabric.csv')
metabric.head()
```

## a)

```python
sns.ecdfplot(metabric, x='Overall Survival (Months)')
plt.title('ECDF Plot of Survival')
plt.show()
```

## b)

```python
sns.ecdfplot(metabric, x='Overall Survival (Months)', hue='Chemotherapy')
plt.title('ECDF Plot of Survival by Chemotherapy Treatment')
plt.show()
```

## c)

Since the `NO` group is shifted more to the right, that group lives longer on average. Therefore, we expect the average patient receiving chemotherapy to live a shorter amount than someone not on chemo.

## d)

We can not tell if chemotherapy is effective purely based on the data. We would need to compare the group receiving chemotherapy to a control group with equivalent diseases who are not receiving chemo.

In real life, we know that chemotherapy is effective. The reason the NO group lives longer is that they are less sick to begin with, and chemo is given to patients who are much sicker. Chemotherapy is helping the YES group live longer than they would with no chemotherapy, even though the YES group is still behind the NO group.
